# **Pareto analysis – 80/20 rule**

```python
product_sales = (
    df.groupby("coffee_name")["money"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

product_sales["revenue_percentage"] = (
    product_sales["money"] /
    product_sales["money"].sum() * 100
)

product_sales["cumulative_percentage"] = (
    product_sales["revenue_percentage"].cumsum()
)

fig, ax1 = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=product_sales,
    x="coffee_name",
    y="money",
    ax=ax1
)

ax2 = ax1.twinx()

ax2.plot(
    product_sales["coffee_name"],
    product_sales["cumulative_percentage"],
    marker="o"
)

ax2.axhline(80, linestyle="--")

ax1.set_title("Pareto Analysis of Coffee Revenue")
ax1.set_xlabel("Coffee")
ax1.set_ylabel("Revenue")
ax2.set_ylabel("Cumulative Revenue (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# **Customer segmentation using RFM analysis**

```python
customer_df = df[df["card"].notna()].copy()

snapshot_date = (
    customer_df["datetime"].max()
    + pd.Timedelta(days=1)
)

rfm = customer_df.groupby("card").agg(
    recency=("datetime", lambda x:
             (snapshot_date - x.max()).days),
    frequency=("datetime", "count"),
    monetary=("money", "sum")
)

print(rfm.head())
```

# **Create customer segments**

```python
rfm["R_score"] = pd.qcut(
    rfm["recency"].rank(method="first"),
    4,
    labels=[4, 3, 2, 1]
)

rfm["F_score"] = pd.qcut(
    rfm["frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)

rfm["M_score"] = pd.qcut(
    rfm["monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)

rfm["RFM_score"] = (
    rfm["R_score"].astype(str) +
    rfm["F_score"].astype(str) +
    rfm["M_score"].astype(str)
)

print(rfm.head())
```

# **Visualize customer segments**

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=rfm,
    x="frequency",
    y="monetary",
    size="recency",
    hue="R_score",
    palette="viridis"
)

plt.title("Customer Segmentation: RFM Analysis")
plt.xlabel("Purchase Frequency")
plt.ylabel("Total Spending")

plt.show()
```

# **Average time between customer purchases**

```python
customer_df = customer_df.sort_values(
    ["card", "datetime"]
)

customer_df["previous_purchase"] = (
    customer_df.groupby("card")["datetime"]
    .shift(1)
)

customer_df["days_between"] = (
    customer_df["datetime"] -
    customer_df["previous_purchase"]
).dt.total_seconds() / 86400

purchase_intervals = customer_df.dropna(
    subset=["days_between"]
)

plt.figure(figsize=(10, 6))

sns.histplot(
    data=purchase_intervals,
    x="days_between",
    bins=30,
    kde=True
)

plt.title("Days Between Customer Purchases")
plt.xlabel("Days Between Purchases")
plt.ylabel("Frequency")

plt.show()
```

# **Sales forecasting preparation**

```python
daily_sales = (
    df.groupby("date")["money"]
    .sum()
    .reset_index()
    .sort_values("date")
)

daily_sales = daily_sales.set_index("date")

daily_sales["lag_1"] = (
    daily_sales["money"].shift(1)
)

daily_sales["lag_7"] = (
    daily_sales["money"].shift(7)
)

daily_sales["rolling_mean_7"] = (
    daily_sales["money"]
    .shift(1)
    .rolling(7)
    .mean()
)

daily_sales["day_of_week"] = (
    daily_sales.index.dayofweek
)

daily_sales["month"] = (
    daily_sales.index.month
)

print(daily_sales.head(15))
```
